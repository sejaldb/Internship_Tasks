## Q1. Find the Frequency of Each Word in a Sentence

**Pseudocode**

Split the sentence into words. For each word, increase its count in a dictionary or add it with a count of one.

## Q2. Find the Frequency of Each Character in a Word

**Pseudocode**

Start with an empty dictionary. Visit each character and count how often it occurs.

## Q3. Find the First Non-Repeating Character in a Word

**Pseudocode**

Count all characters first. Scan the word again and return the first character whose count is one. If none exists, return None.

## Q4. Check Whether Two Strings Are Anagrams

**Pseudocode**

Count the characters in both strings. They are anagrams if both frequency dictionaries are equal.

## Q5. Find the First Repeating Character in a String

**Pseudocode**

Keep a set of characters already seen. Return the first character that is already in the set; otherwise add it to the set.

## Q6. Move All Zeros to the End of a List

**Pseudocode**

Copy non-zero values to a result list while counting zeros. Append that many zeros after scanning the list.

## Q7. Find the Second Highest Distinct Number in a List

**Pseudocode**

Remove duplicates, sort the remaining values from largest to smallest, and return the second value if it exists. Otherwise return None.

## Q8. Remove Duplicates Without Changing the Original Order

**Pseudocode**

Scan the list from left to right. Add an item to the result only if it has not been seen before.

## Q9. Find the Missing Number from 1 to N

**Pseudocode**

Calculate the sum of all integers from 1 to N. Subtract the sum of the given list from this expected sum.

## Q10. Find Two Numbers Whose Sum Equals a Target

**Pseudocode**

Scan numbers from left to right and keep earlier numbers in a set. For each number, check whether its complement has already appeared. Return the first pair found, or an empty list.

## Q11. Find the Number Closest to Zero

**Pseudocode**

Choose the number with the smallest absolute value. If absolute values tie, choose the positive number.

## Q12. Find Common Elements Between Two Lists

**Pseudocode**

Make a set from the second list. Scan the first list and add values found in that set, skipping values already added.

## Q13. Rotate a List to the Right by K Positions

**Pseudocode**

If the list is empty, return an empty list. Reduce K modulo the list length, then move the last K items to the front.

## Q14. Find the Maximum Sum of Two Consecutive Elements

**Pseudocode**

Calculate each adjacent pair's sum and return the largest sum.

## Q15. Find All Pairs Whose Sum Equals a Target

**Pseudocode**

Check every pair of distinct positions. Keep pairs whose sum equals the target, sort the two values within each pair, remove duplicate pairs, and sort the final list.
